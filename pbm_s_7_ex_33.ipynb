**33.** Implement the gradient algorithm with Armijo step size rule (Algorithm 6.1 in the lecture notes). Use as input data the starting vector $x^0$, the parameter for the stopping criterion $\varepsilon$, and the parameters $\sigma$ and $\beta$ for the determination of the Armijo step size. The sequence $x^0, x^1, x^2, \dots$ containing the iteration history should be returned and the points $(x^0, f(x^0)), (x^1, f(x^1)), (x^2, f(x^2)), \dots$ should be plotted on the graph of the function $f$.

The implemented algorithm should be tested for the following functions and input data values:

* **(a)** $f(x) = \cos(x)$, $x^0 = 0.5$, $\varepsilon = 10^{-3}$, $\sigma = 10^{-2}$ and $\beta = 0.5$.
* **(b)** $f(x_1, x_2) = (x_1^2 + x_2 - 11)^2 + (x_1 + x_2^2 - 7)^2$ (Himmelblau function), $\varepsilon = 10^{-1}$, $\sigma = 10^{-2}$, $\beta = 0.5$ and $x^0 = (-0.27, -0.91)^T$, $x^0 = (-0.271, -0.91)^T$, $x^0 = (-0.25, -1.1)^T$ and $x^0 = (-0.25, -1)^T$.
* **(c)** $f(x_1, x_2) = 6x_1^2 - 6x_1x_2 + 2x_2^2 + x_1 + x_2 + 1$, $x^0 = (1, 2)^T$, $\varepsilon = 10^{-2}$, $\sigma = 10^{-2}$ and $\beta = 0.5$.

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

````python
# define function for case (a), (b), (c) and their gradients
#case(a)
def f_a(x):
    return np.cos(x)

def grad_f_a(x):
    return -np.sin(x)

#case (b)
# himmelblau function

def f_b(v):
    x,y = v
    return (x**2 + y - 11)**2 + (x +y**2 -7)**2

def grad_f_b(v):
    x, y = v
    return np.array([
        4*(x**2 + y -11)*x + 2*(x + y**2 -7),
        2*(x**2 + y -11) + 4*(x + y**2 - 7)*y
    ])

# case (c)
def f_c(v):
    x, y = v
    return 6*x**2 - 6*x*y + 2*y**2 + x + y + 1

def grad_f_c(v):
    x, y = v
    return np.array([
        12*x - 6*y + 1,
        4*y - 6*x + 1
    ])
````

````python
def gradient_algorithm_armijo(f, grad_f, x0, epsilon, sigma, beta):
    x_k = x0
    k = 0  # num of iterations
    vectors = []
    vectors.append(x0)

    while np.linalg.norm(grad_f(x_k)) > epsilon:
        d_k = - grad_f(x_k)

        t_k = 1
        while f(x_k + t_k*d_k) > f(x_k) + sigma*t_k*np.dot(grad_f(x_k),d_k):
            t_k *= beta
        
        x_k = x_k + t_k * d_k
        k += 1
        vectors.append(x_k)

    return x_k, vectors, k
````

````python
# case (a)
x0 = 0.5
epsilon = 1e-3
sigma = 1e-2
beta = 0.5

x_a, vectors_a, num_it_a = gradient_algorithm_armijo(f_a, grad_f_a, x0, epsilon, sigma, beta)
print(f'Starting point:', x0)
print(f'Last approximation:', x_a)
print(f'Number of iterations:', num_it_a)
print(f'Implementation history:', np.array(vectors_a, dtype=float))
````

````python
vec_a_arr = np.array(vectors_a, dtype=float)

x_curve = np.linspace(0, 3.5, 400)
y_curve = f_a(x_curve)

plt.figure(figsize=(8, 5))
plt.plot(x_curve, y_curve, label='$f(x) = \\cos(x)$', color='blue', alpha=0.7)

plt.scatter(vec_a_arr, f_a(vec_a_arr), color='red', s=50, zorder=5, label='Iteration Points $(x^k, f(x^k))$')
plt.plot(vec_a_arr, f_a(vec_a_arr), color='red', alpha=0.8, label='Optimization Path')

plt.title("Optimization path")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
````

````python
# case (b)
x1 = [-0.27, -0.91]
x2 = [-0.271, -0.91]
x3 = [-0.25, -1.1]
x4 = [-0.25, -1]

starting_points = [x1, x2, x3, x4]

epsilon = 1e-1
sigma = 1e-2
beta = 0.5

number_of_iterations = []
x_ks = []


for x in starting_points:
    x_k, vectors_a, num_it = gradient_algorithm_armijo(f_b, grad_f_b, x, epsilon, sigma, beta)
    x_ks.append(x_k)
    number_of_iterations.append(num_it)


data = {
    "Starting point": starting_points,
    "Last x_k": x_ks,
    "Number of iterations": number_of_iterations
}

df = pd.DataFrame(data)
df
````

Himmelblau's function has four local minima (ref: [Wikipedia](https://en.wikipedia.org/wiki/Himmelblau%27s_function)):
* $f(3, 2) = 0$
* $f(-2.805118, 3.131312) = 0$
* $f(-3.779310, -3.283186) = 0$
* $f(3.584428, -1.848126) = 0$

````python

x_vals = np.linspace(-5, 5, 400)
y_vals = np.linspace(-5, 5, 400)
X, Y = np.meshgrid(x_vals, y_vals)
Z = (X**2 + Y - 11)**2 + (X + Y**2 - 7)**2

plt.figure(figsize=(10, 8))
levels = np.logspace(0, 4.5, 35)
plt.contour(X, Y, Z, levels=levels, cmap='jet', alpha=0.7)

# loop trough starting pts
for i, x_start in enumerate(starting_points):
    x_end, vectors_b, num_it_b = gradient_algorithm_armijo(f_b, grad_f_b, x_start, 0.1, 1e-2, 0.5)
    vec_b_arr = np.array(vectors_b, dtype=float)
    
    # paths
    plt.plot(vec_b_arr[:, 0], vec_b_arr[:, 1], marker='.', markersize=5, label=f'Start {i+1} ({num_it_b} iters)')
    plt.scatter(vec_b_arr[0, 0], vec_b_arr[0, 1],marker='^', s=40, zorder=5)

plt.title("Optimization path finding minima of the cosine", fontsize=12)
plt.xlabel("$x_1$", fontsize=11)
plt.ylabel("$x_2$", fontsize=11)
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
````

````python
# case (c)
x0 = [1, 2]
epsilon = 1e-2
sigma = 1e-2
beta = 0.5

x_c, vectors_c, num_it_c = gradient_algorithm_armijo(f_c, grad_f_c, x0, epsilon, sigma, beta)
print(f'Starting point:', x0)
print(f'Last approximation:', x_c)
print(f'Number of iterations:', num_it_c)
print(f'Implementation history:', np.array(vectors_c, dtype=float))
````
